<h1>Dictionaries</h1>

```python
#dictionaries store items using key:values e.g "name": "emmanuel
profile = {
    "name": "darrel",
    "age": 37,
    "hobbies": ["sleeping", "playing" , "football", "reading books"],
    "height": 3
}
```

```python
#Dictionary methods 
# --1 accessing items
print(profile["name"])
```

```python
profile.get("name")
```

```python
#returns nothing when empty
profile.get("country")
```

```python
# --2 Adding items to a dictionary
profile["country"] = "USA"
profile
```

```python
profile["country"] = "afghanistan"
profile
```

```python
# --3 deleting
del profile["hobbies"]
profile.pop("age")
profile
```

```python
profile
```

```python
#---4 Getting keys
print(profile.keys())
```

```python
#----5 getting values 
print(profile.values())
```

```python
#---- 6 items
print(profile.items())
```

```python
settings = {"theme":"light"}
#print(settings.get("font","ARIAL")) ITreturns arial when font is absent
settings.setdefault("notifications","pending")
print(settings)
```

```python
#----7 updating
new_font = {"font": "sans_serif"}
settings.update(new_font)
settings
```

```python
#----  looping over a dictionary
# looping by keys
for key in settings.keys():
    print(key)
```

```python
for value in settings.values():
    print(value)
```

```python
for item in settings.items():
    print(item)
```

```python
sentence = "God is the answer and they are saying that they are not tired. God is watching"
#.split converts into list
list_of_words = sentence.split()
# .join reverses
#s2= " ".join(list_of_words)
word_count = {}
for word in list_of_words:
    if word in word_count.keys():
        word_count[word]+=1
    else:
        word_count[word] = 1
word_count



```
