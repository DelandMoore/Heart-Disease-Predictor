```python
scores = [85,92,78,90,66,74]
sum_total = 0
count = 0
for number in range(len(scores)):
    sum_total = sum_total + scores[number]
    count+=1
average = sum_total/count
print(f"Sum of all the numbers: {sum_total}\n")
print( f"The average is: {average}")
```

<h5>exercise 17</h5>

```python
population = {'Tokyo': 37400068, 'Delhi': 30290936, 'Shanghai': 27058480, 'Cairo': 20900604}
max_pop=population['Tokyo']
min_pop=population['Tokyo']

max_city='Tokyo'
min_city='Tokyo'
for city,pop in population.items():
    if pop > max_pop:
        max_pop = pop
        min_city = city
    if pop < min_pop:
        min_pop = pop
        min_city = city
        
print(f"Maximum population: {max_city} with {max_pop}")
print(f"Minimum population: {min_city} with {min_pop}")
```
