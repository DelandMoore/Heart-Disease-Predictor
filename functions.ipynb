```python
def greet(name, language = "English"):
    """say hello to someone in choosen language"""
    if language == "French":
        return f"Bonjour, {name}!"
    return f"Hello, {name}!"
print(greet("Deland"))
print(greet("Deland","French"))
```

<h1>keyword arguments</h1>

```python
def prof(c,g):
    p = g-c
    if p<0:
        return "loss"
    else:
        return "gain"
prof(g=50,c=10)
```

<h4>*args and **kwargs</h4>

```python
# *args: this parameter takes in any number of parameters
#* makes in a tuple
def total(*values):
    t = sum(values)
    return t
total(2,6,4,2,2,2,2,2)
    
```

```python
# **kwargs: this parameter takes in any number of keyword arguments (e.g a = 10)
#             and stores in a dictionary
def computer_specs(**specs):
    return specs
computer_specs(color="pink", ram_size="16GB")
```

```python
def computer_specs2(**specs):
    for i in specs.values():
        print(i)
    return specs
computer_specs2(x=4, y=7)
```
