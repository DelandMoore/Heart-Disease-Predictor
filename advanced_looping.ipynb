<h1>Advanced Looping</h1>

```python
teams = ["barca", "arsenal", "man u"]
pos = [1,2,7]
for item in zip(teams,pos):
    print(item)
    print(type(item))
```

```python
for team,position in zip(teams,pos):
    print(f" {team.upper()} will finish the league at position number: {position} this season\n")
```

<h2>Enumerate</h2>

```python
#prints index and value at that position
for item in enumerate (teams):
    print(item)
```

```python
 for position,element in enumerate(teams):
     print(f"{position} : {element}")
     
```
